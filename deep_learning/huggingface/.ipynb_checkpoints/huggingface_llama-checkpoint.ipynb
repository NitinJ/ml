```python
# Load mistralai/Mistral-7B-v0.3 pretrained model.

from transformers import AutoModelForCausalLM, AutoTokenizer
from transformers import pipeline as pipeline_hf

model_id = 'mistralai/Mistral-7B-v0.3'
model = AutoModelForCausalLM.from_pretrained(model_id, token=True)
tokenizer = AutoTokenizer.from_pretrained(model_id, token=True)

pipeline = pipeline_hf('text-generation', model=model, tokenizer=tokenizer)
output = pipeline("The capital of France", top_k=5)
print(output)
```
