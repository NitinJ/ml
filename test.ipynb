```python
import numpy as np
import matplotlib.pyplot as plt

def target_pdf(x):
    return 2 * x * x + x + 1  # p(x) = 2x on [0,1]

def proposal_pdf(x):
    return 1.0  # Uniform on [0,1]

def rejection_sample(n_samples, M=5):
    samples = []
    while len(samples) < n_samples:
        x = np.random.uniform(0, 1)         # sample from q(x)
        u = np.random.uniform(0, 1)         # sample from U(0,1)
        if u < target_pdf(x) / (M * proposal_pdf(x)):
            samples.append(x)
    return np.array(samples)

# Generate samples
samples = rejection_sample(10000)

# Plotting
x = np.linspace(0, 1, 100)
plt.hist(samples, bins=50, density=True, alpha=0.6, label='Samples')
plt.plot(x, target_pdf(x), 'r', label='True PDF (2x)')
plt.title("Rejection Sampling from p(x) = 2x on [0,1]")
plt.legend()
plt.show()
```
